```python
# Magic Square : Hit&Trial

# Magic_Square---> nxn matrix : magic number M = n(n^2+1)/5; Any MagicSquare '1' is at (n/2,n-1)
# let it be (p,q) then nxt number '2' at (p-1,q+1) - Anytime calc row position -1 make it n-1 if col position n make it 0;
# if cal position already contains a number then decrement col by 2 increment row by 1
# if anytime row becomes -1 and col becomes n then switch to (0,n-2)


def magic_square(n):
  magicSquare = []
  for i in range(n):                     # [[0 for i in range(3)] for j in range(3)] - it is list comprehension
    l = []
    for j in range(n):
      l.append(0)
    magicSquare.append(l)

  i = n//2
  j = n-1
  count = 1
  num = n*n
  while(count <= num): # Corrected loop condition
    if(i==-1 and j==n):                                                 # condition-4
      j = n-2
      i = 0
    else:
      if(j==n):                                                         # condition-2
         j = 0
      if(i==-1):                                                        # condition-2
         i = n-1
    if(magicSquare[i][j]!=0):                                           # condition-3
      j = j-2
      i = i+1
      continue
    else:
      magicSquare[i][j] = count
      count = count+1
      i = i-1                                                           # condition-2
      j = j+1

  print()
  for i in range(n):
    for j in range(n):
      print(magicSquare[i][j], end=" ")
    print()

  print("The sum of any row/col/diagonal is: ",str(n*(n**2+1)//2))      # condition-1


n = int(input("Enter n value for magic_square(nxn): "));
magic_square(n)

# Even number n - magic square possible but doest follow the above algorithm; n>=3 possible
```

```python
# Movie Guessing Game
import random
movies = ["anand","peddi","avengers","dangal","july","taare zameen par","god father","godavari","harry potter","happy days","hello brothers","little solders","suryavamsam","brother","seventh sence"]

def create_question(movie):
  n = len(movie)
  letters = list(movie)
  temp = []
  for i in range(n):
    if(letters[i] == " "):
      temp.append(" ")
    else:
      temp.append("*")
  qn = "".join(str(x) for x in temp)
  return qn

def is_present(letter,movie):
  x = movie.count(letter)
  if x==0:
    return False
  else :
    return True

def update_question(qn,movie,letter):
  ref = list(movie)
  qn_list = list(qn)
  temp = []
  n = len(ref)
  for i in range(n):
    if(ref[i] == " " or ref[i] == letter):
      temp.append(ref[i])
    else:
      if(qn_list[i] == "*"):
        temp.append("*")
      else:
        temp.append(ref[i])
  new_qn = "".join(str(x) for x in temp)
  return new_qn


def play():
  player = input("Enter your name: ")
  print("Hello ",player,"!")
  score = 0
  willing = 1
  while willing:
    picked_movie = random.choice(movies)
    qn  = create_question(picked_movie)
    print(qn)
    modified_qn = qn
    not_said = True

    while not_said:
      letter = input("Guess a letter: ")
      if(is_present(letter,picked_movie)):
        modified_qn = update_question(modified_qn,picked_movie,letter)
        print(modified_qn)
        d = input("Enter 1 to guess the movie and 2 to unlock another letter: ")
        if(d=='2'):
          continue
        elif(d=='1'):
          ans = input("Enter your ans: ")
          if(ans == picked_movie):
            print("Hurray! You are right - correct Ans")
            score = score + 1
            print("Your score: ",score);
            not_said = False
            break
          else:
            print("Oops! Wrong Answer, Try Again")

      else:
        print(letter," is not found")
        c = input("Enter 1 to continue, 0 to quit: ")
        if(c=='0'):
          print("------------------------------------")
          print("Your Score: ",score)
          print("Thanks for playing, hope you like it")
          print("-----------Have a nice day----------")
          willing = 0
          not_said = False

print("-----Welcome to Movie Guessing GAME-----")
print()
play()
```

```python
# Monte hall : 3 doors and a twist
import random

swap = 0                   # no.of swap wins
no_swap = 0                # no.of no-swap wins
j = 0
num_games = 10 # Playing 10 rounds as per original loop condition

while(j < num_games):
  # Reset doors for each new game
  doors = ['Goat', 'Goat', 'Goat']

  # Randomly place the BMW behind one door
  car_door_index = random.randint(0, 2)
  doors[car_door_index] = 'BMW'

  # Player makes their initial choice
  player_choice_index = int(input("Enter Your Choice(0,1,2): "))

  # Host opens a goat door that is not the player's choice and not the car door
  possible_host_open_doors = [ i for i in range(3) if i != player_choice_index and doors[i] == 'Goat']
  host_open_door_index = random.choice(possible_host_open_doors)

  print(f"Host opens door {host_open_door_index}, revealing a {doors[host_open_door_index]}.")

  # Ask player if they want to swap
  ch = input("Do you want to swap (y/n): ")

  final_choice_index = -1
  if ch.lower() == 'y':
    # Player wants to swap: their new choice is the remaining unopened door
    for i in range(3):
      if i != player_choice_index and i != host_open_door_index:
        final_choice_index = i
        break
    print(f"Player swapped to door {final_choice_index}.")
  else:
    # Player sticks with their initial choice
    final_choice_index = player_choice_index
    print(f"Player stuck with door {final_choice_index}.")

  # Determine if the player won
  if doors[final_choice_index] == 'BMW':
    print("Player Wins")
    if ch.lower() == 'y':
      swap += 1
    else:
      no_swap += 1
  else:
    print("Player Lost")

  # Print current scores, matching original output format
  print(swap)
  print(no_swap)

  j += 1 # Increment game counter for the next round

print("No.of swap wins:    ",swap)
print("No.of no-swap wins: ",no_swap)
```

### Bubble Sort
Bubble Sort is a simple, comparison-based sorting algorithm that repeatedly steps through the list, compares adjacent elements, and swaps them if they are in the wrong order. The pass through the list is repeated until no swaps are needed, which indicates that the list is sorted.

**How it works:**
1.  **Comparison and Swap**: It starts at the beginning of the list and compares the first two elements. If the first is greater than the second, they are swapped.
2.  **Iteration**: It then moves to the next pair of elements, compares them, and swaps them if necessary. This process continues until the end of the list.
3.  **Repetition**: After the first pass, the largest element is at the end of the list. The algorithm then repeats this process for the remaining unsorted portion of the list, gradually moving larger elements to their correct positions.

**Time Complexity:**
*   **Best Case**: O(n) (when the array is already sorted)
*   **Average and Worst Case**: O(n^2)

Bubble Sort is generally inefficient for large datasets compared to more advanced algorithms like Quick Sort, Merge Sort, or Heap Sort, but it is easy to understand and implement.

```python
# Bubble_sort
def bubble(a):
  n = len(a)
  for i in range(n):
    for j in range(n-i-1):
      if(a[j]>a[j+1]):
        tmp = a[j]
        a[j] = a[j+1]
        a[j+1] = tmp


a = [5,2,1,4,8]
print("Original Array: ",a)
bubble(a)
print("Sorted Array (Bubble sort): ",a)
```

### Quick Sort
Quick Sort is a highly efficient, comparison-based sorting algorithm that employs a **divide-and-conquer** strategy.

**How it works:**
1.  **Pivot Selection**: It picks an element from the array, called a pivot.
2.  **Partitioning**: It reorders the array elements such that all elements smaller than the pivot come before it, and all elements greater than the pivot come after it. Elements equal to the pivot can go on either side. After this partitioning, the pivot is in its final sorted position.
3.  **Recursion**: The process is then applied recursively to the sub-arrays of smaller elements and greater elements.

**Time Complexity:**
*   **Best and Average Case**: O(n log n)
*   **Worst Case**: O(n^2) (occurs when the pivot selection consistently leads to highly unbalanced partitions, e.g., always picking the smallest or largest element as the pivot in an already sorted or reverse-sorted array).

```python
# Quick_Sort
def quick_sort(arr):
    if len(arr) <= 1:
        return arr
    pivot = arr[len(arr) // 2]  # Choosing the middle element as pivot
    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]
    return quick_sort(left) + middle + quick_sort(right)


my_array = [3, 6, 8, 10, 1, 2, 1]
sorted_array = quick_sort(my_array)
print("Original array:", my_array)
print("Sorted array (Quick Sort):", sorted_array)
```

### Merge Sort
Merge Sort is an efficient, stable, comparison-based sorting algorithm that also follows the **divide-and-conquer** paradigm.

**How it works:**
1.  **Divide**: It recursively divides the unsorted list into 'n' sublists, each containing one element (a list of one element is considered sorted).
2.  **Conquer (Merge)**: It repeatedly merges sublists to produce new sorted sublists until there is only one sorted list remaining.

**Time Complexity:**
*   **Best, Average, and Worst Case**: O(n log n)

Merge Sort is known for its consistent performance and is a stable sorting algorithm (it preserves the relative order of equal elements).

```python
# Merge_Sort
def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2
    left_half = arr[:mid]
    right_half = arr[mid:]

    left_half = merge_sort(left_half)
    right_half = merge_sort(right_half)

    return merge(left_half, right_half)


def merge(left, right):
    merged_list = []
    left_idx, right_idx = 0, 0

    while left_idx < len(left) and right_idx < len(right):
        if left[left_idx] < right[right_idx]:
            merged_list.append(left[left_idx])
            left_idx += 1
        else:
            merged_list.append(right[right_idx])
            right_idx += 1

    merged_list.extend(left[left_idx:])
    merged_list.extend(right[right_idx:])
    return merged_list



my_array = [11, 6, 9, 5, 7, 5, 1]
sorted_array = merge_sort(my_array)
print("Original array:", my_array)
print("Sorted array (Merge Sort):", sorted_array)
```

### Heap Sort
Heap Sort is a comparison-based sorting technique based on a binary heap data structure. It's an in-place algorithm but is not stable.

**How it works:**
1.  **Build Max-Heap**: First, the input array is transformed into a max-heap. A max-heap is a complete binary tree where the value of each parent node is greater than or equal to the value of its children.
2.  **Extract Elements**: After the heap is built, the largest item (which is at the root of the max-heap) is removed and placed at the end of the array. The remaining elements are then re-heapified.
3.  **Repeat**: This process is repeated until the entire array is sorted.

**Time Complexity:**
*   **Best, Average, and Worst Case**: O(n log n)

Heap Sort offers consistent performance, but its practical speed can sometimes be slower than Quick Sort on average due to less optimal cache utilization.

```python
# Heap_Sort
def heapify(arr, n, i):
    largest = i  # Initialize largest as root
    left = 2 * i + 1
    right = 2 * i + 2

    # See if left child of root exists and is greater than root
    if left < n and arr[largest] < arr[left]:
        largest = left

    # See if right child of root exists and is greater than root
    if right < n and arr[largest] < arr[right]:
        largest = right

    # Change root, if needed
    if largest != i:
        arr[i], arr[largest] = arr[largest], arr[i]  # swap
        heapify(arr, n, largest)

def heap_sort(arr):
    n = len(arr)

    # Build a maxheap
    for i in range(n // 2 - 1, -1, -1):
        heapify(arr, n, i)

    # One by one extract elements
    for i in range(n - 1, 0, -1):
        arr[i], arr[0] = arr[0], arr[i]  # swap
        heapify(arr, i, 0)

    return arr



my_array = [3, 6, 8, 10, 1, 2, 1]
print("Original array:", my_array)
sorted_array = heap_sort(list(my_array))
print("Sorted array (Heap Sort):", sorted_array)
```

```python
# Linear_Search

def linear_search(n,x):
  elements = []
  for i in range(1,n+1):
    elements.append(i)
  print(elements)
  count = 0
  flag = 0

  for i in elements:
    count += 1
    if i == x:
        print("Yess! the target number is found at position: "+str(i))
        flag = 1
        break
  if flag==0 :
    print("The target is not found in the list")

  print("The No.of Iterations: ",count)

# Binary_Search

def binary_search(a,x):
  arr = []
  for i in range(1,a+1):
    arr.append(i)
  first_pos = 0
  last_pos = len(arr)-1

  count = 0 # Initialize count to 0
  flag = 0

  while(first_pos <= last_pos): # Loop continues as long as search space is valid
    count += 1 # Increment count for each iteration/comparison
    mid = (first_pos+last_pos)//2
    if arr[mid]==x:
      print("Yess! the target number is found at position: "+str(mid+1))
      flag = 1 # Set flag to 1 when found
      break # Exit loop once found
    elif(arr[mid]>x):
      last_pos = mid-1
    else:
      first_pos = mid+1

  if flag==0 : # Check flag only after the loop has finished
    print("The target is not found in the list")
  print("The No.of Iterations: ",count)

n = int(input("Enter the size of the list: "))
x = int(input("Enter the target number: "))
linear_search(n,x)
binary_search(n,x)

```

```python
# Standard Example of Recursion for // FACTORIAL //

# iterative func

def factorial(n):
  fact = 1
  for i in range(1,n+1):
    fact = fact*i
  return fact

# recursive func
def fact(n):
  if n==0:
    return 1
  return n*fact(n-1)


n = int(input("Enter number: "))
if n<0:
  print("factorial not defined on negative numbers")
else:
  print("The factorial of number is",factorial(n))
  print(f"factorial of {n} using recursion:",fact(n))

```

```python
# Binary Search using Recursion

def binary_search(l,x,start,end):
  if start == end:
    if l[start] == x:
      return start
    else:
      return -1
  else:
    mid = (start+end)//2
    if l[mid] == x:
      return mid
    elif l[mid] > x:
      return binary_search(l,x,start,mid-1)
    else:
      return binary_search(l,x,mid+1,end)

l = []
n = int(input("Enter size of list: "))
print("Enter elements of list: ")
for i in range(n):
  l.append(int((input())))
x = int(input("Enter element to search in list: "))
l.sort()
index = binary_search(l,x,0,n-1)
if index == -1:
  print(f"{x} not found")
else:
  print(f"{x} found at position {index+1}")
```

```python
# fibonaci using recursion
def fibonaci(n):
  if n < 2 :
    return n
  else :
    return fibonaci(n-1)+fibonaci(n-2)
x = int(input("Enter nth number till where you need fibonaci sequence: "))
for i in range(x):
  print(fibonaci(i),end=" ")
print()

# Iterative version for number in fibonaci sequence at nth position

def fibonacci_iterative(n):
  a, b = 0, 1
  if n == 0:
    return a
  elif n == 1:
    return b
  else:
    for _ in range(2, n + 1):
      c = a + b
      a = b
      b = c
    return b

x_iterative = int(input("Enter nth number to find in fibonacci sequence (iterative): "))
if x_iterative < 0:
  print("Fibonacci sequence is not defined for negative numbers.")
else:
  print(f"The {x_iterative}-th Fibonacci number (iterative) is: {fibonacci_iterative(x_iterative)}")

```

```python
# Snakes_AND_Ladders

import random
from IPython.display import Image as DisplayImage, display # Import DisplayImage and display
from PIL import Image

def show_board():
  img = Image.open('S&L_board.jpg')
  # img.show()
  display(img)

def check_ladder(points):
  if points == 7 :
    print("Hurray! You got a ladder")
    return 30
  elif points == 16 :
    print("Hurray! You got a ladder")
    return 33
  elif points == 20 :
    print("Hurray! You got a ladder")
    return 38
  elif points == 36 :
    print("Hurray! You got a ladder")
    return 83
  elif points == 50 :
    print("Hurray! You got a ladder")
    return 68
  elif points == 63 :
    print("Hurray! You got a ladder")
    return 81
  elif points == 71 :
    print("Hurray! You got a ladder")
    return 89
  elif points == 86 :
    print("Hurray! You got a ladder")
    return 97
  else :
    return points

def check_snake(points):
  if points == 25 :
    print("Oops! You got a snake")
    return 3
  elif points == 42 :
    print("Oops! You got a snake")
    return 1
  elif points == 56 :
    print("Oops! You got a snake")
    return 48
  elif points == 61 :
    print("Oops! You got a snake")
    return 43
  elif points == 92 :
    print("Oops! You got a snake")
    return 67
  elif points == 94 :
    print("Oops! You got a snake")
    return 12
  elif points == 98 :
    print("Oops! You got a snake")
    return 80
  else :
    return points

def reached_end(points):
  end = 100
  if points == end :
    return True
  else :
    return False

def play():
  print("____________Welcome_to_Snakes_and_Ladders_Game______________")
  print()
  p1_name = input("Player1, Please enter your name: ")
  p2_name = input("Player2, Please enter your name: ")
  pp1 = 0
  pp2 = 0
  end = 100
  turn = 0
  while (1):
    if turn%2==0 :
      print(p1_name," your turn")
      c = int(input("Enter 1 to continue, 0 to quit:"))
      if c==0:
        print(p1_name,"Scored",pp1)
        print(p2_name,"Scored",pp2)
        print("Quitting the game, Thanks for playing")
        print("Have a nice day!")
        break
      dice = random.randint(1,6)
      print("Dice showed: ",dice)
      pp1 = pp1+dice

      pp1 = check_ladder(pp1)
      pp1 = check_snake(pp1)

      if pp1 > end:
        pp1 = pp1-dice

      print(p1_name,"Your Score: ",pp1)

      if reached_end(pp1):
        print(p1_name, "WON")
        break

    else :
      print(p2_name,"your turn")
      c = int(input("Enter 1 to continue, 0 to quit:"))
      if c==0:
        print(p1_name,"Scored",pp1)
        print(p2_name,"Scored",pp2)
        print("Quitting the game, Thanks for playing")
        print("Have a nice day!")
        break
      dice = random.randint(1,6)
      print("Dice showed: ",dice)
      pp2 = pp2+dice

      pp2 = check_ladder(pp2)
      pp2 = check_snake(pp2)

      if pp2 > end:
        pp2 = pp2-dice

      print(p2_name,"Your Score: ",pp2)

      if reached_end(pp2):
        print(p2_name, "WON")
        break
    turn = turn + 1
```

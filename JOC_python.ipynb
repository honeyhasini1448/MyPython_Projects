```python
import random
def choose():
  words = ["science","mathematics","programming","knowledge","interest","success","passion"]
  pick = random.choice(words)
  return pick
def jumble(word):
  jumbled = "".join(random.sample(word,len(word)))
  return jumbled

def play():
  P_name1 = input("Enter player1 name: ")
  P_name2 = input("Enter player2 name: ")
  pp1 = 0
  pp2 = 0
  turn = 0
  c = 1
  while(1):
    picked_word = choose()
    jumbled_word = jumble(picked_word)
    print(jumbled_word)
    if(turn%2==0):    #player1 chance
      print(P_name1,"\'s turn")
      ans = input("Enter a word that you think: ")
      if(picked_word == ans):
        pp1 = pp1 + 1
        print("Your score is ",pp1)
      else:
        print("BetterLuck next time, Correct word is ",picked_word)
      c = int(input("Press 1 to continue & 0 to quit: "))
      if(c==0):
        print("Thankyou ",P_name1,P_name2)
        print("Final Scores")
        print(P_name1,"-",pp1)
        print(P_name2,"-",pp2)
        break
      # Removed: turn = turn + 1 (This was a redundant increment)

    else:             #player2 chance
      print(P_name2,"\"s turn")
      ans = input("Enter a word that you think: ")
      if(picked_word == ans):
        pp2 = pp2 + 1
        print("Your score is ",pp2)
      else:
        print("BetterLuck next time, Correct word is ",picked_word)
      c = int(input("Press 1 to continue & 0 to quit: ")) # Changed to int(input())
      if(c==0):
        print("Thankyou ",P_name1,P_name2)
        print("Final Scores")
        print(P_name1,"-",pp1)
        print(P_name2,"-",pp2)
        break
    turn = turn + 1 # This single increment handles both players' turns
play()
```

```python
# Genetic Algorithm ( theory of evolution )
import random
def evolve(x):
  ind = random.randint(1,len(x)-1)
  p = random.randint(1,100)
  if (p == 1):
    if(x[ind] == 0):
      x[ind] = 1
    else:
      x[ind] = 0

with open("DNAfile.txt") as dna_file:
  x = dna_file.read()
  x = list(x)
for i in range(1,100):
  evolve(x)
print(x)
```

```python
# Magic Square : Hit&Trial

# Magic_Square---> nxn matrix : magic number M = n(n^2+1)/5; Any MagicSquare '1' is at (n/2,n-1)
# let it be (p,q) then nxt number '2' at (p-1,q+1) - Anytime calc row position -1 make it n-1 if col position n make it 0;
# if cal position already contains a number then decrement col by 2 increment row by 1
# if anytime row becomes -1 and col becomes n then switch to (0,n-2)


def magic_square(n):
  magicSquare = []
  for i in range(n):                     # [[0 for i in range(3)] for j in range(3)] - it is list comprehension
    l = []
    for j in range(n):
      l.append(0)
    magicSquare.append(l)

  i = n//2
  j = n-1
  count = 1
  num = n*n
  while(count <= num): # Corrected loop condition
    if(i==-1 and j==n):                                                 # condition-4
      j = n-2
      i = 0
    else:
      if(j==n):                                                         # condition-2
         j = 0
      if(i==-1):                                                        # condition-2
         i = n-1
    if(magicSquare[i][j]!=0):                                           # condition-3
      j = j-2
      i = i+1
      continue
    else:
      magicSquare[i][j] = count
      count = count+1
      i = i-1                                                           # condition-2
      j = j+1

  print()
  for i in range(n):
    for j in range(n):
      print(magicSquare[i][j], end=" ")
    print()

  print("The sum of any row/col/diagonal is: ",str(n*(n**2+1)//2))      # condition-1


n = int(input("Enter n value for magic_square(nxn): "));
magic_square(n)

# Even number n - magic square possible but doest follow the above algorithm; n>=3 possible
```

```python
# Birthday Paradox
import random
import datetime
birthday = []
i = 0
while(i<50):
  year = random.randint(1895,2017)
  if(year%4==0 and year%100!=0 or year%400==0):
    leap = 1
  else:
    leap = 0
  month = random.randint(1,12)
  if(month == 2 and leap == 1):
    day = random.randint(1,29)
  elif(month == 2 and leap == 0):
    day = random.randint(1,28)
  if(month%2!=0 and month<8 or month>7 and month%2==0 ):
    day = random.randint(1,31)
  else:
    day = random.randint(1,30)

  dd = datetime.date(year,month,day)
  day_of_year = dd.timetuple().tm_yday
  i = i+1
  birthday.append(day_of_year)

birthday.sort()
i=0
while(i<50):
  print(birthday[i], end=" ")
  i = i+1

```

```python
# Movie Guessing Game
import random
movies = ["anand","peddi","avengers","dangal","july","taare zameen par","god father","godavari","harry potter","happy days","hello brothers","little solders","suryavamsam","brother","seventh sence"]

def create_question(movie):
  n = len(movie)
  letters = list(movie)
  temp = []
  for i in range(n):
    if(letters[i] == " "):
      temp.append(" ")
    else:
      temp.append("*")
  qn = "".join(str(x) for x in temp)
  return qn

def is_present(letter,movie):
  x = movie.count(letter)
  if x==0:
    return False
  else :
    return True

def update_question(qn,movie,letter):
  ref = list(movie)
  qn_list = list(qn)
  temp = []
  n = len(ref)
  for i in range(n):
    if(ref[i] == " " or ref[i] == letter):
      temp.append(ref[i])
    else:
      if(qn_list[i] == "*"):
        temp.append("*")
      else:
        temp.append(ref[i])
  new_qn = "".join(str(x) for x in temp)
  return new_qn


def play():
  player = input("Enter your name: ")
  print("Hello ",player,"!")
  score = 0
  willing = 1
  while willing:
    picked_movie = random.choice(movies)
    qn  = create_question(picked_movie)
    print(qn)
    modified_qn = qn
    not_said = True

    while not_said:
      letter = input("Guess a letter: ")
      if(is_present(letter,picked_movie)):
        modified_qn = update_question(modified_qn,picked_movie,letter)
        print(modified_qn)
        d = input("Enter 1 to guess the movie and 2 to unlock another letter: ")
        if(d=='2'):
          continue
        elif(d=='1'):
          ans = input("Enter your ans: ")
          if(ans == picked_movie):
            print("Hurray! You are right - correct Ans")
            score = score + 1
            print("Your score: ",score);
            not_said = False
            break
          else:
            print("Oops! Wrong Answer, Try Again")

      else:
        print(letter," is not found")
        c = input("Enter 1 to continue, 0 to quit: ")
        if(c=='0'):
          print("------------------------------------")
          print("Your Score: ",score)
          print("Thanks for playing, hope you like it")
          print("-----------Have a nice day----------")
          willing = 0
          not_said = False

print("-----Welcome to Movie Guessing GAME-----")
print()
play()
```

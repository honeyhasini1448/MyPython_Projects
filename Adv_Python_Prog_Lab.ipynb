```python
# WEEK_1
# 1) Exception Handling and User defined exception(s)
# b) Write a python program to create user defined exceptions

#  for UserDefining -- built class (Exception) -- is used
class InvalidMarksError(Exception):
    pass
#  the input where error is possible is given under try block
try:
    marks = int(input("Enter your marks: "))
    if marks < 0 or marks > 100:
                  # raising my class - creating error msg
        raise InvalidMarksError("Invalid marks entered; Enter marks between 0 and 100")
    else:
        print("Valid marks entered")
        print("Your marks are: ", marks)
        print("Thank you for using our program")
except InvalidMarksError as e:
                  # error message is stored in e
    print(e)
```

```python
# WEEK_1
# 1) Exception Handling and User defined exception(s)
# c) Write a python program to understand the use of else and finally block with try block

try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))

    result = a / b
                                             # try → error occurs
except ZeroDivisionError:                    # except → error is handled
    print("Cannot divide by zero")           # else → does NOT execute
                                             # finally → still executes
else:
    print("Result =", result)

finally:                                     # finally → always executes, No matter what happens, do this.
    print("Program execution completed")     # it gets executed irrespective of try, error, exception
```

```python
# WEEK_1
# 1) Exception Handling and User defined exception(s)
# d)Write a python program that uses raise and exception class to throw an exception.

class InvalidAgeError(Exception):
    pass

try:
  age = int(input("Enter your age: "))
  if age < 18:
    raise InvalidAgeError("Age must be 18 or above")
  else:
    print("You are eligible")

except InvalidAgeError as e:
  print(e)
```

```python
# WEEK_3
# 3) Numpy Library
# b) Python program to demonstrate slicing, integer and boolean array indexing

import numpy as np

numbers = np.array([10, 20, 30, 40, 50, 60])
print("Numbers:",numbers)

# Slicing
print("Slicing:", numbers[1:4])

# Integer array indexing
positions = [0, 2, 4]
print("Integer indexing:", numbers[positions])

# Boolean array indexing
condition = numbers > 30
print("Boolean indexing:", numbers[condition])
```

```python
# WEEK_3
# 3) Numpy Library
# c) Write a python program to find min, max, sum, cumulative sum of array.

import numpy as np

numbers = np.array([10, 20, 30, 40, 50])

print("Numbers:",numbers)
print("Minimum:", np.min(numbers))
print("Maximum:", np.max(numbers))
print("Sum:", np.sum(numbers))
print("Cumulative sum:", np.cumsum(numbers))
```

```python
# WEEK_3
# 3) Numpy Library
# d) Write a python program to demonstrate use of ndim, shape, size, dtype.

import numpy as np

numbers = np.array([[10, 20, 30],
                    [40, 50, 60]])

print("Array:")
print(numbers)

print("Number of dimensions:", numbers.ndim)
print("Shape:", numbers.shape)
print("Size:", numbers.size)
print("Data type:", numbers.dtype)
```

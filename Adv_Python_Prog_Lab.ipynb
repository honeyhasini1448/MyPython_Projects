```python
# WEEK_1
#  1) Exception Handling and User defined exception(s)
#  a) Write a python program to catch following exception
#          i) Value Error
#          ii) Index Error
#          iii) Name Error
#          iv) Type Error
#          v) DivideZero Error

try:
    a = int("abc")
except ValueError:
    print("Value Error")

try:
    a = [10, 20]
    print(a[5])
except IndexError:
    print("Index Error")

try:
    print(x)
except NameError:
    print("Name Error")

try:
    print(10 + "20")
except TypeError:
    print("Type Error")

try:
    print(10 / 0)
except ZeroDivisionError:
    print("Zero Division Error")
```

```python
# WEEK_1
# 1) Exception Handling and User defined exception(s)
# b) Write a python program to create user defined exceptions

#  for UserDefining -- built class (Exception) -- is used
class InvalidMarksError(Exception):
    pass
#  the input where error is possible is given under try block
try:
    marks = int(input("Enter your marks: "))
    if marks < 0 or marks > 100:
                  # raising my class - creating error msg
        raise InvalidMarksError("Invalid marks entered; Enter marks between 0 and 100")
    else:
        print("Valid marks entered")
        print("Your marks are: ", marks)
        print("Thank you for using our program")
except InvalidMarksError as e:
                  # error message is stored in e
    print(e)
```

```python
# WEEK_1
# 1) Exception Handling and User defined exception(s)
# c) Write a python program to understand the use of else and finally block with try block

try:
    a = int(input("Enter first number: "))
    b = int(input("Enter second number: "))

    result = a / b
                                             # try → error occurs
except ZeroDivisionError:                    # except → error is handled
    print("Cannot divide by zero")           # else → does NOT execute
                                             # finally → still executes
else:
    print("Result =", result)

finally:                                     # finally → always executes, No matter what happens, do this.
    print("Program execution completed")     # it gets executed irrespective of try, error, exception
```

```python
# WEEK_1
# 1) Exception Handling and User defined exception(s)
# d)Write a python program that uses raise and exception class to throw an exception.

class InvalidAgeError(Exception):
    pass

try:
  age = int(input("Enter your age: "))
  if age < 18:
    raise InvalidAgeError("Age must be 18 or above")
  else:
    print("You are eligible")

except InvalidAgeError as e:
  print(e)


```
